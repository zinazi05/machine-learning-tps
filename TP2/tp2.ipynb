```python
import numpy as np
import pandas as pd
import matplotlib
```

```python
data = {
    "Student":     ["S1", "S2", "S3", "S4", "S5", "S6"],
    "Mathematics": [12, 15, 8, 17, 10, 14],
    "Statistics":  [11, 14, 9, 16, 12, 15],
    "Programming": [14, 16, 10, 18, 11, 13],
    "Algorithms":  [13, 15, 9, 17, 10, 14],
}
```

```python
df = pd.DataFrame(data).set_index("Student")
```

```python
print("Shape (students, variables):", df.shape)
print(df.describe())
```

```python
X = df.values.astype(float)
mean = X.mean(axis=0)
std = X.std(axis=0, ddof=1)
```

```python
Z = (X - mean) / std
Z_df = pd.DataFrame(Z, index=df.index, columns=df.columns)
```

```python
print("Standardized data:\n", Z_df)
print("mean of Z :", Z.mean(axis=0))
print("std  of Z :", Z.std(axis=0, ddof=1))
```

```python
C = np.corrcoef(Z, rowvar=False)
print("Corelation matrix of standardized data:\n",
      pd.DataFrame(C, index=df.columns, columns=df.columns))
```

```python
eigvals, eigvecs = np.linalg.eigh(C)
print("Raw eigenvalues ", eigvals)
print("Raw eigenvectors \n", eigvecs)
```

```python
idx = np.argsort(eigvals)[::-1]
eigvals = eigvals[idx]
eigvecs = eigvecs[:, idx]
```

```python
print("Sorted eigenvalues:", eigvals)
print("Sorted eigenvectors",
      pd.DataFrame(eigvecs))
```

```python
print("Sorted eigenvectors (columns = PC1..PC4):\n",
      pd.DataFrame(eigvecs, index=df.columns,
                   columns=[f"PC{i+1}" for i in range(4)]))
print("Sum of eigenvalues (= number of variables):", eigvals.sum())
```

```python
W = eigvecs[:, :2]
print("Projection matrix W (4 x 2):\n", W)
```

```python
scores = Z @ W           
print(scores)
```

```python
ratio = eigvals / eigvals.sum()
cumulative = np.cumsum(ratio)
print(ratio)
print(cumulative)
```
